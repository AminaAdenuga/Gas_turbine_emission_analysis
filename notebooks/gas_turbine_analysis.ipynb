```python
from pathlib import Path

figures_folder = Path("../figures")
figures_folder.mkdir(parents=True, exist_ok=True)
```

```python
plt.figure(figsize=(12,7))

sns.histplot(
    data=gas_turbine_df,
    x='carbon_monoxide',
    bins=20,
    kde = True
)

plt.title('Distribution Of Carbon Monoxide Emissions')
plt.xlabel('Carbon Monoxide')
plt.ylabel('Frequency/Count')

plt.savefig(
    figures_folder / "carbon_monoxide_histplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

RELATIONSHIPS USING HEATMAPS

```python
correlation_matrix = gas_turbine_df.corr(numeric_only =True)
plt.figure(figsize=(12,9))
sns.heatmap(
correlation_matrix,
annot = True,
cmap ='coolwarm',
fmt ='.2f'
)
plt.savefig(
    figures_folder / "correlation_matrix_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
# Calculate correlations between all numerical variables
correlation_matrix = gas_turbine_df.corr(numeric_only=True)

# Rank relationships with our three target variables
targets = [
    "carbon_monoxide",
    "nitrogen_oxides",
    "turbine_energy_yield"
]

for target in targets:
    print(f"\nStrongest relationships with {target}:")

    display(
        correlation_matrix[target]
        .drop(target)
        .sort_values(key=abs, ascending=False)
        .round(3)
    )
```

```python
from pathlib import Path

# Create the processed data folder if it doesn't exist
output_folder = Path("../data/processed")
output_folder.mkdir(parents=True, exist_ok=True)

# Save cleaned dataset
gas_turbine_df.to_csv(
    output_folder / "gas_turbine_cleaned.csv",
    index=False
)

# Save correlation matrix
correlation_matrix.to_csv(
    output_folder / "gas_turbine_correlation_matrix.csv"
)

print("Project datasets saved successfully!")
```
